## 2.1.3 — Kiểm tra và xử lý dữ liệu lương


### 1. Xem các trường liên quan đến lương


```python
salary_columns = [
    "Luong_Min",
    "Luong_Max",
    "Luong_TB",
    "salary_status",
    "salary_currency_orig",
    "salary_period_assumed",
    "salary_source",
]

display(df[salary_columns].head(10))

display(
    df["salary_status"]
    .value_counts(dropna=False)
    .rename_axis("trang_thai")
    .reset_index(name="so_tin")
)
```

| Trạng thái     | Ý nghĩa                               |
| -------------- | ------------------------------------- |
| `negotiable`   | Cần thương lượng                      |
| `closed_range` | Có đủ hai cận lương                   |
| `missing`      | Thiếu thông tin lương                 |
| `max_only`     | Chỉ có mức tối đa                     |
| `hidden`       | Lương bị ẩn                           |
| `min_only`     | Chỉ có mức tối thiểu                  |
| `negotiable`   | Thỏa thuận                            |
| `unknown`      | Chưa xác định được                    |
| `invalid`      | Quy trình trước đánh dấu không hợp lệ |


-> Không gộp tất cả thành “thỏa thuận”, vì thiếu dữ liệu và nhà tuyển dụng chủ động ghi thỏa thuận là hai trường hợp khác nhau.


### 2. Kiểm tra các cột lương có thực sự là số


```python
numeric_salary_columns = ["Luong_Min", "Luong_Max", "Luong_TB"]

# Chuyển sang số trong một bảng riêng, giữ nguyên cột ban đầu
salary_numeric = df[numeric_salary_columns].apply(pd.to_numeric, errors="coerce")

# Những ô ban đầu có dữ liệu nhưng không chuyển được thành số
parse_errors = df[numeric_salary_columns].notna() & salary_numeric.isna()

print("Số ô không chuyển được thành số:")
display(parse_errors.sum().to_frame("so_o_loi"))

display(df.loc[parse_errors.any(axis=1), ["job_uid", "title"] + numeric_salary_columns])
```

-> Không có lỗi chuyển đổi


### 3. Kiểm tra tính hợp lệ của hai cận

Một khoảng lương cần thỏa mãn:

- Cận được cung cấp phải là số hữu hạn và lớn hơn 0.
- Nếu có đủ hai cận thì Luong_Min ≤ Luong_Max.
- Thiếu một cận không tự động là lỗi.


```python
salary_min = salary_numeric["Luong_Min"]
salary_max = salary_numeric["Luong_Max"]

has_min = salary_min.notna()
has_max = salary_max.notna()
has_both = has_min & has_max

invalid_min = has_min & (~np.isfinite(salary_min) | salary_min.le(0))

invalid_max = has_max & (~np.isfinite(salary_max) | salary_max.le(0))

reversed_bounds = has_both & salary_min.gt(salary_max)

bounds_parse_error = parse_errors[["Luong_Min", "Luong_Max"]].any(axis=1)

df["salary_invalid_bounds"] = (
    invalid_min | invalid_max | reversed_bounds | bounds_parse_error
)

print("Số tin có cận lương không hợp lệ:", df["salary_invalid_bounds"].sum())

display(
    df.loc[
        df["salary_invalid_bounds"],
        ["job_uid", "title", "Luong_Min", "Luong_Max", "salary_status"],
    ]
)
```

-> Không tự hoán đổi hai cận nếu thấy min > max: lỗi có thể nằm ở việc trích số hoặc đơn vị, không chỉ ở thứ tự.


### 4. Kiểm tra nhãn lương có khớp với dữ liệu số


| Nhãn `salary_status`                            | Dữ liệu thực tế                     |
| ----------------------------------------------- | ----------------------------------- |
| `closed_range`                                  | Thiếu một hoặc cả hai cận           |
| `min_only`                                      | Không có `Luong_Min`, có cả hai cận |
| `max_only`                                      | Không có `Luong_Max`, có cả hai cận |
| `negotiable` hoặc `hidden`                      | Có ít nhất một cận                  |
| `missing` hoặc `unknown`                        | Có ít nhất một cận                  |
| `invalid`                                       | Hai cận nhìn có vẻ hợp lệ           |
| Nhãn bị thiếu, sai chính tả hoặc ngoài danh mục | Bất kỳ                              |


```python
df["salary_bounds_status"] = "no_numeric"

df.loc[has_min & ~has_max, "salary_bounds_status"] = "min_only"

df.loc[~has_min & has_max, "salary_bounds_status"] = "max_only"

df.loc[has_both, "salary_bounds_status"] = "closed_range"

numeric_statuses = ["closed_range", "min_only", "max_only"]

# Nhãn nói có lương số nhưng hai cận không khớp
numeric_status_conflict = df["salary_status"].isin(numeric_statuses) & df[
    "salary_status"
].ne(df["salary_bounds_status"])

# Nhãn nói không có lương số nhưng vẫn xuất hiện số
nonnumeric_status_conflict = ~df["salary_status"].isin(numeric_statuses) & (
    has_min | has_max
)

df["salary_status_conflict"] = numeric_status_conflict | nonnumeric_status_conflict

display(pd.crosstab(df["salary_status"], df["salary_bounds_status"]))

print("Số tin có nhãn lương không nhất quán:", df["salary_status_conflict"].sum())
```

### 5. Gắn cờ các giả định chưa được kiểm chứng


```python
# Cờ chu kỳ lương được giả định
df["salary_period_unverified"] = df["salary_period_assumed"].eq(True)

# Dùng Luong_TB có sẵn, đã chuyển sang số ở bước 2
salary_mean = salary_numeric["Luong_TB"]

# Đủ điều kiện phân tích lương sơ bộ
df["salary_eda_eligible"] = (
    df["salary_bounds_status"].eq("closed_range")
    & ~df["salary_invalid_bounds"]
    & ~df["salary_status_conflict"]
    & salary_mean.notna()
    & np.isfinite(salary_mean)
    & salary_mean.gt(0)
    & salary_mean.ge(salary_numeric["Luong_Min"])
    & salary_mean.le(salary_numeric["Luong_Max"])
)

# Tập bảo thủ: chỉ lấy tin không giả định chu kỳ lương
df["salary_eda_conservative"] = df["salary_eda_eligible"] & df[
    "salary_period_assumed"
].eq(False)
```

### 6. Cập nhật tập IT và tạo các tập phân tích lương


```python
df_it = df.loc[df["scope_status"].eq("included_provisional")].copy()

df_salary = df_it.loc[df_it["salary_eda_eligible"]].copy()

df_salary_conservative = df_it.loc[df_it["salary_eda_conservative"]].copy()

print(f"Tổng tin IT dùng sơ bộ: {len(df_it):,}")
print(f"Tin đủ hai cận lương hợp lệ: {len(df_salary):,}")
print(f"Tin thuộc tập lương bảo thủ: {len(df_salary_conservative):,}")

print("Tỷ lệ có đủ hai cận lương:", f"{len(df_salary) / len(df_it) * 100:.2f}%")
```

-> Như vậy, chỉ khoảng 24% tập IT sơ bộ có đủ hai cận để tính trung điểm. Khi làm EDA, ta phải ghi rõ số mẫu; kết quả lương có thể chịu ảnh hưởng bởi việc nguồn nào hoặc cấp bậc nào thường công khai lương.


```python
display(
    df_salary[
        [
            "title",
            "Luong_Min",
            "Luong_Max",
            "Luong_TB",
            "salary_currency_orig",
            "salary_period_assumed",
        ]
    ].head(10)
)
```

## 2.1.4 — Chuẩn hóa địa điểm, kinh nghiệm và cấp bậc


### 1. Chuyển cities thành danh sách địa điểm


```python
import json
import re

# Chuyển chuỗi JSON thành danh sách
cities_parsed = df["cities"].fillna("[]").map(json.loads)

# Kiểm tra cấu trúc: danh sách chứa các chuỗi
assert cities_parsed.map(
    lambda xs: (isinstance(xs, list) and all(isinstance(x, str) for x in xs))
).all()

# Bỏ khoảng trắng, phần tử rỗng và địa điểm lặp
df["cities_clean"] = cities_parsed.map(
    lambda xs: list(dict.fromkeys(x.strip() for x in xs if x.strip()))
)

df["n_cities"] = df["cities_clean"].map(len)

display(df[["city", "cities", "cities_clean", "n_cities"]].head(10))
```

### 3. Kiểm tra khoảng kinh nghiệm


```python
experience_columns = ["exp_min_years", "exp_max_years"]

# Nếu có giá trị không chuyển được thành số, pandas sẽ báo lỗi
df[experience_columns] = df[experience_columns].apply(pd.to_numeric)

exp_min = df["exp_min_years"]
exp_max = df["exp_max_years"]

df["experience_review"] = (
    exp_min.lt(0)
    | exp_max.lt(0)
    | exp_min.gt(exp_max)
    | (exp_min.notna() & ~np.isfinite(exp_min))
    | (exp_max.notna() & ~np.isfinite(exp_max))
)

print("Tin có khoảng kinh nghiệm cần kiểm tra:", df["experience_review"].sum())

display(df.loc[df["experience_review"], ["job_uid", "title"] + experience_columns])
```

Không điền kinh nghiệm thiếu bằng 0. Không biết yêu cầu kinh nghiệm khác với việc công việc chấp nhận người chưa có kinh nghiệm.

không phát hiện cận âm, vô hạn hoặc min lớn hơn max.


### 4. Tạo nhóm kinh nghiệm để thống kê

Ta nhóm theo số năm kinh nghiệm tối thiểu được yêu cầu, phục vụ EDA.


```python
df["experience_group"] = (
    pd.cut(
        exp_min.mask(df["experience_review"]),
        bins=[0, 1, 3, 5, np.inf],
        labels=["0–<1 năm", "1–<3 năm", "3–<5 năm", "≥5 năm"],
        right=False,
    )
    .astype("string")
    .fillna("Unknown")
)

display(df["experience_group"].value_counts().to_frame("so_tin"))
```

Nhóm kinh nghiệm không đồng nghĩa với cấp bậc. Ta không tự kết luận tất cả tin yêu cầu 3 năm đều là Middle, vì cách đặt cấp bậc khác nhau giữa doanh nghiệp.


### 5. Giữ nhãn cấp bậc có sẵn và tách Intern

Trước hết, giữ `seniority` làm cơ sở; giá trị thiếu được ghi rõ là `Unknown`. Lưu thêm nguồn xác định nhãn.


```python
df["seniority_clean"] = (
    df["seniority"].astype("string").str.strip().replace("", pd.NA).fillna("Unknown")
)

df["seniority_basis"] = df["seniority_source"].fillna("Unknown")

title = df["title"].fillna("")

# Nhận diện thực tập sinh khi tiêu đề ghi rõ
intern_title = title.str.contains(
    r"\bintern(?:ship)?\b|\btrainee\b|\btts\b|thực\s*tập", case=False, na=False
)

df.loc[intern_title, "seniority_clean"] = "Intern"
df.loc[intern_title, "seniority_basis"] = "title_intern"
```

Không dùng riêng cờ `is_intern` để đổi tất cả sang Intern, vì dữ liệu trước đó có thể gắn cờ này cho cộng tác viên. Những trường hợp đó cần đối chiếu:


```python
df["intern_flag_review"] = df["is_intern"].eq(True) & ~intern_title

display(
    df.loc[
        df["intern_flag_review"], ["title", "is_intern", "seniority", "seniority_clean"]
    ]
)
```

Cờ này có nghĩa cần xem lại bằng chứng, chưa kết luận nhãn cũ sai.


Tiếp theo, nhận diện cách viết nhiều cấp bậc rõ ràng như `Junior/Senior` hoặc `Middle/Senior`:


```python
level_pattern = r"(?:intern(?:ship)?|fresher|junior|middle|mid|senior|lead(?:er)?)"

multi_pattern = (
    rf"\b{level_pattern}" rf"\s*(?:/|,|&|-|\bor\b|\band\b)\s*" rf"{level_pattern}\b"
)

multi_level = title.str.contains(multi_pattern, case=False, na=False)

df.loc[multi_level, "seniority_clean"] = "Multi-level"
df.loc[multi_level, "seniority_basis"] = "title_multiple_levels"

display(df.loc[multi_level, ["title", "seniority", "seniority_clean"]].head(10))
```

## 2.1.5 — Làm sạch JD và chuẩn hóa danh sách kỹ năng.


### 1. Làm sạch văn bản JD


```python
import html
import json
import re
import unicodedata


def clean_jd(value):
    if pd.isna(value):
        return ""

    text = unicodedata.normalize("NFC", html.unescape(str(value)))

    # Bỏ các thẻ HTML đơn giản
    text = re.sub(r"</?[A-Za-z][^>]*>", " ", text)

    # Bỏ ký tự vô hình và gộp khoảng trắng
    text = text.replace("\u200b", "")
    return re.sub(r"\s+", " ", text).strip()


df["description_clean"] = df["description"].map(clean_jd)

df["has_jd_clean"] = df["description_clean"].ne("")

display(df[["description", "description_clean"]].head(3))

print("Tin thiếu JD:", (~df["has_jd_clean"]).sum())
```

| Thao tác              | Ví dụ                                    |
| --------------------- | ---------------------------------------- |
| `html.unescape()`     | `&amp;` → `&`                            |
| Chuẩn hóa Unicode NFC | Thống nhất cách biểu diễn dấu tiếng Việt |
| Bỏ thẻ HTML           | `<p>Python</p>` → `Python`               |
| Gộp khoảng trắng      | Nhiều dấu cách/xuống dòng → một dấu cách |

Không xóa toàn bộ dấu câu, vì sẽ làm biến dạng C++, C#, .NET, CI/CD. Cột description gốc vẫn được giữ để đối chiếu.


### 2. Chuyển `tech`, `keywords` thành danh sách Python


```python
for column in ["tech", "keywords"]:
    parsed = df[column].fillna("[]").map(json.loads)

    assert parsed.map(
        lambda xs: (isinstance(xs, list) and all(isinstance(x, str) for x in xs))
    ).all(), f"Cột {column} có cấu trúc không hợp lệ."

    df[f"{column}_list"] = parsed

display(df[["tech", "tech_list", "keywords", "keywords_list"]].head(3))
```

### 3. Chuẩn hóa một số tên tương đương và loại trùng


```python
skill_aliases = {
    "powerbi": "Power BI",
    "power bi": "Power BI",
    "nodejs": "Node.js",
    "node.js": "Node.js",
    "postgres": "PostgreSQL",
    "postgresql": "PostgreSQL",
    "gcp": "Google Cloud",
    "sklearn": "scikit-learn",
}


def normalize_skills(items):
    return sorted(
        {
            skill_aliases.get(item.strip().casefold(), item.strip())
            for item in items
            if item.strip()
        }
    )


for column in ["tech_list", "keywords_list"]:
    df[column] = df[column].map(normalize_skills)
```

Những tên chưa có trong bảng tương đương được giữ nguyên. Không gộp SQL Server thành SQL, hoặc React Native thành React, vì chúng không phải cùng một tên kỹ năng.


### 4. Gộp kỹ năng và tách các nhóm từ khóa khác


keywords chứa cả công nghệ, khái niệm chung và lĩnh vực ứng dụng. Ta tách chúng để ma trận kỹ năng có ý nghĩa rõ hơn.


```python
# Gộp hai danh sách, mỗi từ khóa chỉ xuất hiện một lần
df["skills"] = df.apply(
    lambda row: sorted(set(row["tech_list"] + row["keywords_list"])), axis=1
)

broad_concepts = {"AI", "Cloud"}
application_domains = {"Fintech", "E-commerce"}

# Giữ riêng khái niệm chung
df["broad_concepts"] = df["skills"].map(lambda xs: sorted(set(xs) & broad_concepts))

# Giữ riêng lĩnh vực ứng dụng
df["application_domains"] = df["skills"].map(
    lambda xs: sorted(set(xs) & application_domains)
)

# Danh sách dùng cho ma trận kỹ năng chính
df["skills"] = df["skills"].map(
    lambda xs: [x for x in xs if x not in broad_concepts | application_domains]
)

df["n_skills"] = df["skills"].str.len()

display(
    df[["title", "skills", "broad_concepts", "application_domains", "n_skills"]].head(5)
)
```

Đây là một lựa chọn biểu diễn dữ liệu:

- Python, AWS, Machine Learning… được giữ trong danh sách chính.
- AI, Cloud khá rộng nên giữ riêng để tránh lấn át các công nghệ cụ thể.
- Fintech, E-commerce mô tả lĩnh vực ứng dụng.

Các thông tin này vẫn được giữ trong bảng. Ta cũng không tự thêm Cloud khi thấy AWS, vì điều đó có thể tạo ra luật kết hợp đúng do chính quy tắc thêm nhãn.


### 5. Tạo nhánh văn bản bỏ stop words


Danh sách kỹ năng đã được xử lý riêng. Bây giờ mới tạo văn bản phục vụ TF-IDF nếu cần dùng ở bước sau.


```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

vi_stopwords = set(
    "và là của các những cho với được trong một "
    "để về khi từ theo tại vào này đó".split()
)

# Giữ một số từ phủ định
stopwords = (set(ENGLISH_STOP_WORDS) - {"no", "not", "never", "without"}) | vi_stopwords


def remove_stopwords(text):
    tokens = re.findall(r"[.#]?\w[\w.+#/-]*", text.lower())

    return " ".join(
        token for token in tokens if token not in stopwords and not token.isdigit()
    )


df["text_for_tfidf"] = df["description_clean"].map(remove_stopwords)
```

```python
# Chạy kiểm tra nhỏ để bảo đảm không làm hỏng tên công nghệ:
assert clean_jd("<p>C++ &amp; C# / .NET</p>") == "C++ & C# / .NET"

assert normalize_skills(["PowerBI", "Power BI", " C++ "]) == ["C++", "Power BI"]

assert {"c++", "c#", ".net"} <= set(remove_stopwords("C++ C# .NET và the").split())
```
